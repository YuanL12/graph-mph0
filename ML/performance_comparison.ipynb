```python
import sys
sys.path.append('/home/yluo/Documents/rivet-python')
from pyrivet import rivet
#Equivalent to the 'points2.txt' sample data file in the RIVET distribution
# Function-VRips Filtration
points = rivet.PointCloud(
    points = [
        (0.3, 1.5, 2.1), # will appear at 1
        (4.2, 3.8, 4.9), # will appear at 3.2
        (5.6, 2.3, 6),   # will appear at 2
        (2.9, 5.1, 3.3), # will appear at 4
        (3.3, 2.0, 2.5), # will appear at 2.5
        (4.1, 1.1, 2.3), # will appear at 2.4
        (1.1, 1.3, 1.7), # will appear at 2
    ],
    second_param_name='appearance',
    appearance=[1, 3.2, 2, 4, 2.5, 2.4, 2],
    max_dist=3.1 # pyrivet will calculate max_dist if you leave it out
)

# 7 points, 3 dimensions
# 6 function values => 6 grades in one dimension for function values
# 7 points => 7+1=8 grades in another dimension for distance

# Now let's take the Betti information
betti = rivet.betti(points)
```

```python
print("Betti information:")
print("rank")
print(betti.graded_rank)

print("xi_0")
print(betti.xi_0)

print("xi_1")
print(betti.xi_1)

print("xi_2")
print(betti.xi_2)

print("x(row) grades")
print(betti.dimensions.x_grades)
print("y(column) grades")
print(betti.dimensions.y_grades)
```

```python
import pandas as pd
# round the x_grades and y_grades to 2 decimal places and store them in a list
x_grades = [round(grade.numerator/grade.denominator, 2) for grade in betti.dimensions.x_grades]
y_grades = [round(grade.numerator/grade.denominator, 4) for grade in betti.dimensions.y_grades]

# transform the betti into a dataframe with rows and columbs labels as the grades 
betti_df = pd.DataFrame(betti.graded_rank.T, index=x_grades, columns=y_grades)
# add column names to the dataframe
betti_df.columns.name = 'distance'
betti_df.index.name = 'function value'
# convert the value to int
betti_df = betti_df.astype(int)
betti_df
```
